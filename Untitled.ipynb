### IMPORT LIBRARIES

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
print("Libraries imported successfully!")
```

### LOAD DATA

```python
df = pd.read_csv("Traffic_Accident_Dataset.csv")
print("Dataset loaded successfully!")
df.head()
```

### EXPLORE DATA

```python
print("Dataset Shape:")
print(df.shape)
print("\nColumn Names:")
print(df.columns.tolist())
```

```python
print("Data Types:")
print(df.dtypes)
```

```python
print("Missing Values:")
print(df.isnull().sum())
```

```python
print("Number of Duplicate Rows:")
print(df.duplicated().sum())
```

### CLEAN DATA

```python
# Convert Date and Time into appropriate formats
df["Date"] = pd.to_datetime(df["Date"], errors="coerce")
df["Time"] = pd.to_datetime(df["Time"], format="%H:%M:%S",
errors="coerce")
# Remove duplicate rows
df = df.drop_duplicates()
print("Data cleaning completed successfully.")
print("Final dataset shape:", df.shape)
```

```python
print("Missing values after cleaning:")
print(df.isnull().sum())
print("\nDuplicate rows after cleaning:")
print(df.duplicated().sum())
```

### ACCIDENT ANALYSIS BY TIME

```python
df["Hour"] = df["Time"].dt.hour
hourly_accidents = df["Hour"].value_counts().sort_index()
print("Accidents by Hour:")
print(hourly_accidents)
print("\nPeak accident hour:",
hourly_accidents.idxmax(),
"with", hourly_accidents.max(), "accidents.")
```

```python
df["Month"] = df["Date"].dt.to_period("M").astype(str)
monthly_accidents = df.groupby("Month").size()
print("Monthly Accident Counts:")
print(monthly_accidents)
```

### ACCIDENT TREND LINE CHART

```python
plt.figure(figsize=(10, 5))
monthly_accidents.plot(kind="line", marker="o")
plt.title("Monthly Traffic Accident Trend")
plt.xlabel("Month")
plt.ylabel("Number of Accidents")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

### ACCIDENT ANALYSIS BY LOCATION / AREA

```python
area_accidents = df["Location/Area"].value_counts()
print("Accidents by Area:")
print(area_accidents)
```

### AREA-WISE BAR CHART

```python
plt.figure(figsize=(10, 5))
area_accidents.plot(kind="bar")
plt.title("Area-wise Traffic Accidents")
plt.xlabel("Location / Area")
plt.ylabel("Number of Accidents")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### SEVERITY ANALYSIS

```python
severity_counts = df["Severity"].value_counts()
print("Accidents by Severity:")
print(severity_counts)
```

### SEVERITY DISTRIBUTION PLOT

```python
plt.figure(figsize=(8, 5))
severity_counts.plot(kind="bar")
plt.title("Traffic Accident Severity Distribution")
plt.xlabel("Severity")
plt.ylabel("Number of Accidents")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

### ACCIDENT ANALYSIS BY VEHICLE TYPE

```python
vehicle_accidents = df["Vehicle Type"].value_counts()
print("Accidents by Vehicle Type:")
print(vehicle_accidents)
```

### ACCIDENT TYPE ANALYSIS

```python
accident_type_counts = df["Accident Type"].value_counts()
print("Accidents by Accident Type:")
print(accident_type_counts)
```

```python

print("Most common accident type:",
accident_type_counts.idxmax(),
"with", accident_type_counts.max(), "records.")
print("Most common vehicle type:",
vehicle_accidents.idxmax(),
"with", vehicle_accidents.max(), "records.")
print("Most frequent accident area:",
area_accidents.idxmax(),
"with", area_accidents.max(), "records.")
```
